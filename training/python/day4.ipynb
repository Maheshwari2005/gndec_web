```python

# creating a class for student name, age, roll number, marks, and percentage

class Student:
    def __init__(self, name, age, roll_number, marks):
        self.name = name
        self.age = age
        self.roll_number = roll_number
        self.marks = marks
        self.percentage = (marks / 500) * 100

    def __str__(self):
        return f"Name: {self.name}, Age: {self.age}, Roll Number: {self.roll_number}, Marks: {self.marks}, Percentage: {self.percentage}%"
```

```python
# creation of objects for student class

student1 = Student("John", 20, 101, 450)
print(student1)
```

```python

# mymath.py

import math
from statistics import mean, median, mode

# Arithmetic operators
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        raise ValueError("Cannot divide by zero")
    return a / b


# Trigonometric functions
def sine(angle):
    return math.sin(math.radians(angle))

def cosine(angle):
    return math.cos(math.radians(angle))

def tangent(angle):
    return math.tan(math.radians(angle))


# Statistical functions
def find_mean(data):
    return mean(data)

def find_median(data):
    return median(data)

def find_mode(data):
    return mode(data)
                                                                             
```

```python
class Person:
    def __init__(self, name, age, bloodgroup, nationality):
        self.name = name
        self.age = age
        self.bloodgroup = bloodgroup
        self.nationality = nationality

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Blood Group:", self.bloodgroup)
        print("Nationality:", self.nationality)


p1 = Person("Maheshwari", 21, "O+", "Indian")

p1.display()
```

```python
class PERSON:
    def __init__(self, name, age, blood, nationality):
        self.name = name
        self.age = age
        self.blood = blood
        self.nationality = nationality


class STUDENT(PERSON):
    def __init__(self, name, age, blood, nationality,
                 school_college, percentage, section, roll_number):
        super().__init__(name, age, blood, nationality)
        self.school_college = school_college
        self.percentage = percentage
        self.section = section
        self.roll_number = roll_number

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Blood Group:", self.blood)
        print("Nationality:", self.nationality)
        print("School/College:", self.school_college)
        print("Percentage:", self.percentage)
        print("Section:", self.section)
        print("Roll Number:", self.roll_number)


student1 = STUDENT("Vaishali", 21, "O+", "Indian",
                   "ABC College", 85.5, "A", 51)

student1.display()
    
```

```python
class Student:
    # Public data members
    name = "Vaishali"
    dept = "ECE"
    year = "4th Year"
    college = "GNDEC"

    def __init__(self):
        # Private data members
        self.__regno = "3GN23EC051"
        self.__percentage = 85
        self.__sec = "A"
        self.__fathername = "Rajendra"
        self.__address = "Bidar"

    def display(self):
        print("Name:", Student.name)
        print("Dept:", Student.dept)
        print("Year:", Student.year)
        print("College:", Student.college)
        print("Reg No:", self.__regno)
        print("Percentage:", self.__percentage)
        print("Section:", self.__sec)
        print("Father Name:", self.__fathername)
        print("Address:", self.__address)

s = Student()
s.display()
```

```python
from abc import ABC, abstractmethod

class Student(ABC):

    @abstractmethod
    def show(self):
        pass

class Details(Student):

    def show(self):
        print("Name: Maheshwari")
        print("Department: ECE")
        print("College: GNDEC")
        print("Year: 4")

s = Details()
s.show()
```
